```python
### textloader
from langchain_community.document_loaders import TextLoader

loader = TextLoader("../data/raw/text_files/ml.txt", encoding="utf-8")
```

```python
loader.load()
```

```python
### Directory loader
from langchain_community.document_loaders import DirectoryLoader
dir_loader = DirectoryLoader("../data/raw/text_files", glob="**/*.txt", loader_cls=TextLoader, loader_kwargs={'encoding': 'utf-8'})
```

```python
dir_loader
```

```python
documents = dir_loader.load()
```
